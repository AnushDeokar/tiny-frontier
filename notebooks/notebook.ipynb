```python
from tiny_frontier.data.tokenizer import create_bpe_tokenizer

from tiny_frontier.data.tokenizer import save_tokenizer
from tiny_frontier.data.tokenizer import load_tokenizer
from tiny_frontier.data.tokenizer import tokenize_text

from tiny_frontier.data.tinystories import load_tinystories
```

```python
from tiny_frontier.data.tokenizer import train_tokenizer
data = load_tinystories()

tokenizer = create_bpe_tokenizer()
tokenizer = train_tokenizer(tokenizer, data)
tokenizer = save_tokenizer(tokenizer, "data/tokenizer.json")
```

```python
from tiny_frontier.data.tokenizer import decode_text
created_tokenizer = load_tokenizer("data/tokenizer.json")
tokenized_text = tokenize_text(created_tokenizer, "Hello, world!")
tokenized_text_2 = tokenize_text(created_tokenizer, "One day, a little girl named Lily")
tokenized_text_3 = tokenize_text(created_tokenizer, "unbelievable")
tokenized_text_4 = tokenize_text(created_tokenizer, "aaaaaaaaaaaaaaaa")

print("Hello, world!", tokenized_text)
print("One day, a little girl named Lily", tokenized_text_2)
print("unbelievable", tokenized_text_3)
print("aaaaaaaaaaaaaaaa", tokenized_text_4)

print("Decoded text=====")
decoded_text = decode_text(created_tokenizer, tokenized_text)
print(decoded_text)
```

```python
from tiny_frontier.model.model import LlamaModel

def count_parameters(model):
    return sum(p.numel() for p in model.parameters() if p.requires_grad)

model = LlamaModel(
    vocab_size=4096,
    d_model=128,
    n_heads=4,
    d_ff=352,
    n_layers=4,
    max_seq_len=256,
)

print(f"Parameters: {count_parameters(model):,}")
```

```python
import torch
input_ids = torch.randint(0, 4096, (2, 32))

logits = model(input_ids)

print(logits.shape)
```

```python
import torch.nn.functional as F

def lm_loss(logits, targets):
    B, T, V = logits.shape

    return F.cross_entropy(
        logits.view(B * T, V),
        targets.view(B * T),
    )
```
